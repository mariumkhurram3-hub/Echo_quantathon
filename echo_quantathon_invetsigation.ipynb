MODULE 1:

The Gatekeeper

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
sim = AerSimulator()
def run(qc, n=1000): return sim.run(qc, shots=n).result().get_counts()

# Bronze: always 1
qc = QuantumCircuit(1,1); qc.x(0); qc.measure(0,0); print(run(qc))   # {'1':1000}
# Bronze: 50/50
qc = QuantumCircuit(1,1); qc.h(0); qc.measure(0,0); print(run(qc))   # ~500/500
# Silver: H then H -> always 0
qc = QuantumCircuit(1,1); qc.h(0); qc.h(0); qc.measure(0,0); print(run(qc))
# Gold: 000 -> 101
qc = QuantumCircuit(3,3); qc.x(0); qc.x(2); qc.measure([0,1,2],[0,1,2]); print(run(qc))  # {'101':1000}
print(qc.draw())
```

**Bronze:** X gives a bit that is always 1. H gives a bit that is 50/50.
*Proves:* I can make one bit that is certain and one that is random.

**Silver:** H then H always gives 0.
*Proves:* The randomness was never real. The second H cancels the first by interference.

**Gold:** X on qubits 0 and 2 turns 000 into 101.
*Proves:* I can build an exact bit pattern with only X.

MODULE 4:
The Erased Measurement

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

sim = AerSimulator()

# Bronze 1: recoverable (no measurement in the middle)
qc = QuantumCircuit(1, 1)
qc.h(0)
qc.h(0)
qc.measure(0, 0)
print("H, H, measure:", sim.run(qc, shots=1000).result().get_counts())

# Bronze 2: destroyed (measure in the middle)
qc = QuantumCircuit(1, 2)
qc.h(0)
qc.measure(0, 0)
qc.h(0)
qc.measure(0, 1)
print("H, measure, H, measure:", sim.run(qc, shots=1000).result().get_counts())

# Silver 1: X then H then H
qc = QuantumCircuit(1, 1)
qc.x(0)
qc.h(0)
qc.h(0)
qc.measure(0, 0)
print("X, H, H:", sim.run(qc, shots=1000).result().get_counts())

# Silver 2: H then X then H
qc = QuantumCircuit(1, 1)
qc.h(0)
qc.x(0)
qc.h(0)
qc.measure(0, 0)
print("H, X, H:", sim.run(qc, shots=1000).result().get_counts())
print(qc.draw())
```

**Bronze:** H, H, measure always gives 0, so the state is recoverable. With a measurement between the two H gates, the result becomes random, so the state was destroyed.
*Proves:* Measuring in the middle destroys the superposition for good. The log was changed by being observed, not hacked.

**Silver:** X, H, H always gives 1. H, X, H always gives 0.
*Proves:* The "erasing" is interference between the two H gates, not data loss.

**Gold:** To predict a new sequence, track the state step by step: X flips 0 and 1, H switches between the 0/1 basis and the +/- basis, and a measurement makes it a plain 0 or 1.
*Proves:* I can predict results before running them.
